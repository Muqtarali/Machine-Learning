```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)

# Load the dataset from the correct path
df = pd.read_csv('/content/heart.csv')
display(df.head())
```

```python
from sklearn.model_selection import train_test_split

X = df.drop('target', axis=1)
y = df['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler

# Initialize StandardScaler
scaler = StandardScaler()

# Scale the training and testing features
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
lr=LogisticRegression(max_iter=1000) # Increased max_iter for better convergence
dt=DecisionTreeClassifier()
```

```python
lr.fit(X_train_scaled,y_train)
dt.fit(X_train_scaled,y_train)
```

```python
y_pred1 = lr.predict(X_test_scaled)
y_pred2 = dt.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
print("Accuracy of Logistic Regression",accuracy_score(y_test,y_pred1))
print("Accuracy of Decision Trees",accuracy_score(y_test,y_pred2))
```

```python
confusion_matrix(y_test,y_pred1)
```

# classification-metrics-multi for multi class\**bold text**

```python

df1 = pd.read_csv('/content/Iris.csv')
df1.head()
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df1['Species'] = encoder.fit_transform(df1['Species'])
df1.head()
```

```python
from sklearn.model_selection import train_test_split

# Correctly split features (X) and target (y) for df1, dropping the 'Id' column
X = df1.drop(['Id', 'Species'], axis=1)
y = df1['Species']

X_train,X_test,y_train,y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler # Import StandardScaler for Iris data
```

```python
clf1 = LogisticRegression(max_iter=1000) # Increased max_iter for convergence
clf2 = DecisionTreeClassifier()
```

```python
# Initialize StandardScaler for Iris data
scaler_iris = StandardScaler()

# Scale the training and testing features for Iris
X_train_scaled_iris = scaler_iris.fit_transform(X_train)
X_test_scaled_iris = scaler_iris.transform(X_test)

clf1.fit(X_train_scaled_iris,y_train)
clf2.fit(X_train_scaled_iris,y_train)
```

```python
y_pred_mul=clf1.predict(X_test_scaled_iris)
y_pred1_mul=clf2.predict(X_test_scaled_iris)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
print(accuracy_score(y_test,y_pred_mul))
print(accuracy_score(y_test,y_pred1_mul))
```

```python
print("Logistic Regression Confusion Matrix\n")
pd.DataFrame(confusion_matrix(y_test,y_pred1_mul),columns=list(range(0,3)))
```
